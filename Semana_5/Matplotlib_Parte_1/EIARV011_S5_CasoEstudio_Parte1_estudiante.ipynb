## 6. De la muestra reducida a la población completa: tamaño muestral y sesgo de selección

El caso original amplía el análisis al rango completo de edades de la encuesta (18 a 55 años) y muestra que la brecha entre Python y los demás lenguajes, marcada entre los 25 y 35 años, se atenúa fuera de ese tramo.

> **Datos simulados (ya generados para ti).** El archivo original con las 38 edades completas no está disponible en este repositorio. La siguiente celda genera una extensión **simulada**, calibrada para aproximarse a los valores reales entre los 25 y 35 años, y que reproduce el patrón cualitativo descrito en el video. No es necesario que modifiques esta celda de datos.

```python
rng = np.random.default_rng(42)
ages_full = np.arange(18, 56)  # 18 a 55 años

def curva_base(edad):
    # Crecimiento salarial típico con la edad, con saturación después de los 45 años
    return 28000 + 48000 / (1 + np.exp(-0.16 * (edad - 33)))

def prima_lenguaje(edad, amplitud, centro=30, ancho=7):
    # Sobresueldo que se concentra entre los 25 y 35 años y se atenúa fuera de ese rango
    return amplitud * np.exp(-((edad - centro) ** 2) / (2 * ancho ** 2))

dev_y_full = np.round(curva_base(ages_full) + rng.normal(0, 300, size=len(ages_full)))
py_dev_y_full = np.round(curva_base(ages_full) + prima_lenguaje(ages_full, 9000)
                          + rng.normal(0, 300, size=len(ages_full)))
js_dev_y_full = np.round(curva_base(ages_full) + prima_lenguaje(ages_full, 3000)
                          + rng.normal(0, 300, size=len(ages_full)))

print("Datos simulados generados para las edades 18 a 55.")
```
