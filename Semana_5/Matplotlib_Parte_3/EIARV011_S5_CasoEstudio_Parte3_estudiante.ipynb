## 1. Mecánica básica del gráfico de pastel

El gráfico de pastel (`plt.pie()`) responde una pregunta distinta a la de un gráfico de barras: no "¿cuánto vale cada categoría?", sino "¿qué proporción del total representa cada categoría?".

Una particularidad útil de `plt.pie()` es que los valores que recibe **no necesitan sumar 100**: internamente calcula la proporción de cada valor respecto de la suma total de la lista. Así, `[60, 40]` y `[120, 80]` producen exactamente el mismo gráfico, porque ambas listas mantienen la proporción 60 %-40 %.
